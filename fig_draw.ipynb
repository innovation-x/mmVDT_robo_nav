```python
import matplotlib.pyplot as plt
import numpy as np

# 创建一个 64x64 的黑色图像
image = np.zeros((64, 64, 3), dtype=np.uint8)

# 创建一个图形和轴
fig, ax = plt.subplots()

# 显示图像
ax.imshow(image)

# 添加一个绿色小箭头
arrow_start = (32, 62)  # 箭头起点
arrow_end = (45, 40)    # 箭头终点
ax.arrow(arrow_start[0], arrow_start[1], arrow_end[0] - arrow_start[0], arrow_end[1] - arrow_start[1],
         head_width=4, head_length=2, fc='green', ec='green', linewidth=10)

# 去除坐标轴
ax.axis('off')

# 显示图形
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 创建一个 64x64 的高斯噪声图像
mean = 0
std = 1
gaussian_noise = np.random.normal(mean, std, (64, 64))

# 创建一个图形和轴
fig, ax = plt.subplots()

# 显示高斯噪声图像
ax.imshow(gaussian_noise, cmap='gray')

# 去除坐标轴
ax.axis('off')

# 显示图形
plt.show()
```
