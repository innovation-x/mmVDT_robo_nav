```python
import numpy as np
import matplotlib.pyplot as plt

# 定义混合高斯分布的参数
a = [0.4, 0.3, 0.3, 0.15]  # 每个高斯分布的权重
b = [-3.8, -2., 2., 4]       # 每个高斯分布的均值
c = [0.8, 1.0, 1., 0.8]  # 每个高斯分布的标准差
linestyles = ['--', '-.', ':', ':']  # 不同的线型

# 定义 x 的范围
x = np.linspace(-6, 6, 500)

# 计算混合高斯分布
f = np.zeros_like(x)
for ai, bi, ci in zip(a, b, c):
    f += ai * np.exp(-((x - bi) ** 2) / (2 * ci ** 2))

# 绘制图像
plt.figure(figsize=(12, 3))
plt.plot(x, f, label="Mixture of Gaussians", linewidth=5)
for ai, bi, ci, ls in zip(a, b, c, linestyles):
    plt.plot(x, ai * np.exp(-((x - bi) ** 2) / (2 * ci ** 2)), ls, label=f"Gaussian (b={bi}, c={ci})", linewidth=5)
plt.axis('off')

plt.show()
```
